```python
def Grouper(df,x,col):
    if 80 <= df[col].loc[x] <= 90:
        return '80-90'
    elif 50 <= df[col].loc[x] <= 60:
        return '50-60'
    else:
        return 'others'
```

```python
g = df.groupby(lambda x: Grouper(df, x, 'temperature'))
g
```

```python
for key,data in g:
    print('Key: ',key)
    print('Data: ',data)
```
