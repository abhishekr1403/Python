```python
import sqlite3

conn = sqlite3.connect("shop.db")

cursor = conn.cursor()

print("Database is ready and connected!")
```

```python
cursor.execute("""
    CREATE TABLE IF NOT EXISTS reviews (
    id INTEGER PRIMARY KEY,
    text TEXT,
    label TEXT,
    score INTEGER
    )
""")

print("Table created and accesed!")
```

```python
cursor.execute(
    "INSERT INTO reviews (text, label, score) VALUES (?, ?, ?)",
    ("The biryani was amazing!", "positive", 5)
)

conn.commit()

print("One Review Saved!")
```

```python
more_reviews = [
    ("Delivery was very late and food was cold.", "negative", 1),
    ("Good taste, fair price.", "positive", 4),
    ("It was okay, nothing special.", "neutral", 3),
    ("App kept crashing during payment.", "negative", 2),
]

cursor.executemany(
    "INSERT INTO reviews (text, label, score) VALUES (?, ?, ?)",
    more_reviews
)
conn.commit()

print("Added", len(more_reviews), "more reviews.")
```

```python
cursor.execute("SELECT * FROM reviews")
rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
cursor.execute("SELECT text, score FROM reviews WHERE label = ?", ("negative",))
negatives = cursor.fetchall()

print("Negative reviews:")
for text, score in negatives:
    print("-", text, "(score:", score, ")")
```

```python
conn.close()
print("Database Closed!")
```

```python
cursor.execute(
    """CREATE TABLE IF NOT EXISTS products (
       id INTEGER PRIMARY KEY,
       name TEXT,
       price INTEGER
    )
""")

print("Table created and accessed!")
```

```python
prod = [
    ("Tea", 15),
    ("Biscuts", 25),
    ("Cake", 30),
]

cursor.executemany(
    "INSERT INTO products (name, price) VALUES (?, ?)",
    prod
)

conn.commit()
print("Added ", len(prod), "products to the database")
```

```python
cursor.execute("SELECT name, price FROM products")
rows = cursor.fetchall()

print("All Products")
for name,price in rows:
    print(f'{name} : {price}')
```

```python
cursor.execute("SELECT name, price FROM products WHERE price > ? ", (20,))
rows = cursor.fetchall()

print("All Products")
for name,price in rows:
    print(f'{name} : {price}')
```
