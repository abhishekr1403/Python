```python
import requests

url = 'https://api.frankfurter.app/latest?from=USD&to=INR'

response = requests.get(url)

print(response.status_code)
```

```python
data = response.json()

print(type(data))
print(data)
print(f"Today 1 USD = {data['rates']['INR']}")
```

```python
try:
    response = requests.get(
        url = 'https://api.frankfurter.app/latest',
        params={"from": "USD", "to": "INR"}
    )

    if response.status_code == 200: 
        data = response.json()
        print("Rate:", data["rates"]["INR"])
    else:
        print("Something went wrong, Status code:",response.status_code)

except Exception as e:
    print("Could not reach the api.", e)
```

```python
def currency_converter(n):
    try:
        response = requests.get(
        url = 'https://api.frankfurter.app/latest',
        params={"from": "EUR", "to": "INR"}
        )

        if response.status_code == 200: 
            data = response.json()
            rate = data["rates"]["INR"]
            # print(rate)
        else:
            print("Something went wrong, Status code:",response.status_code)

    except Exception as e:
        print("Could not reach the api.", e)
    
    converted_amount = n * rate
    print(f'{n} EUR = {converted_amount} INR') 
```

```python
currency_converter(100)
```
