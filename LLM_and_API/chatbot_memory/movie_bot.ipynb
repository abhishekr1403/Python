```python
from dotenv import load_dotenv
from google import genai
from google.genai import types

load_dotenv()

client = genai.Client()
```

```python
conversation = []

def chat(user_message):
    conversation.append({"role": "user", "parts": [{"text": user_message}]})
    
    try:
        config = types.GenerateContentConfig(
            system_instruction="You are a witty movie buff who loves making puns."
        )
        
        response = client.models.generate_content(
            model ='gemini-3.5-flash-lite',
            contents = conversation,
            config = config,
        )
        reply = response.text

        conversation.append({"role": "model", "parts": [{"text": reply}]})
        return reply
    
    except Exception as e:
        print(f"Something went wrong, {e}")

def reset():
    conversation.clear()
```

```python
chat("I love sci-fi and thrillers, but I can't stand horror movies.")
```

```python
chat("Suggest me a movie to watch tonight.")
```

```python
chat("Why did you pick that one?")
```

```python
print(conversation)
```

```python
reset()
```
