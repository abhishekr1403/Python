## Newer interactions method

```python
from google import genai

client = genai.Client()

interaction = client.interactions.create(
    model="gemini-3.8-flash",
    input="Tell me who invented Samosa in 3 lines"
)
print(interaction.output_text)
```

## Older generate_content method

```python
from google import genai
# from google.genai import types

client = genai.Client()
response = client.models.generate_content(
    model="gemini-3.8-flash",
    contents="Tell me who invented Samosa in 3 lines"
)
print(response.text)
```

```python
client = genai.Client()

def generate_recipe_new(ingradients, cuisine, diet):
    '''
    Generate a recipe using ingradients provided. 
    Cuisine can be "Italian", "Mexican", "Indian", "Any" etc.
    diet can be "Vegan", "Vegeterian", "Jain", "Any" etc.
    '''
    prompt = f'''
    Generate one food receipe using these ingradients: {','.join(ingradients)}
    Receipe should not be more than 100 words
    Cuisine: {cuisine}
    Diet: {diet} 
    '''
    interaction = client.interactions.create(
        model="gemini-3.8-flash",
        input=prompt
    )
    return interaction.output_text
```

```python
recipe = generate_recipe_new(["panner", "butter", "onion", "potato","chicken","salt"],"Indian","Non-Vegetarian")
print(recipe)
```
