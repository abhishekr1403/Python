```python
from dotenv import load_dotenv
from google import genai
from pydantic import BaseModel

load_dotenv()
client = genai.Client()
```

```python
class ReviewAnalysis(BaseModel):
    sentiment: str
    rating: int
    pros: list[str]
    cons: list[str]
    recommended_action: str
    
interaction = client.interactions.create(
    model='gemini-3.5-flash-lite',
    input='''I ordered the wireless earbuds last week. Sound quality is amazing and battery
                lasts all day, but the case feels cheap and it arrived two days late. Probably
                won't buy from this brand again.''',
    response_format={
        "type": "text",
        "mime_type": "application/json",
        "schema": ReviewAnalysis.model_json_schema()
    }
    
)

print(interaction.output_text)
```

```python
result = ReviewAnalysis.model_validate_json(interaction.output_text)

print("------------Customer Review Analyzer-------------")
print(f"Sentiment: {result.sentiment}")
print(f"Rating: {result.rating}")
print('Pros: ', end="")
for p in result.pros:
    print(p,", ", end="")
print("")
print('Cons: ', end="")
for c in result.cons:
    print(c,", ", end="")
print("")
print(f"Recommendation: {result.recommended_action}")
```
